## Objetivo
Importar el dataset de clientes de E-commerce, perfilar su estructura, aplicar una regla matemática de negocio mediante computación vectorial con NumPy, segmentar información con `iloc` y comparar la persistencia de los resultados en CSV y Excel.


## Paso 1 — Preparación del entorno e importación de librerías

La guía solicita cargar NumPy, Pandas y `os` como motores de trabajo para la actividad.


```python
import numpy as np
import pandas as pd
import os
```

## Paso 2 — Carga del dataset

Antes de ejecutar la siguiente celda en Google Colab, sube el archivo **`ecommerce_customer_data.csv`** a la sección de archivos temporales del entorno.


```python
# Carga inicial controlando delimitadores y codificación estándar
df_clientes = pd.read_csv(
    'ecommerce_customer_data.csv',
    sep=',',
    encoding='utf-8'
)

# Inspección de las dimensiones de la matriz de datos
print(f"Dimensiones del dataset cargado: {df_clientes.shape}")

# Vista rápida
df_clientes.head()
```

## Paso 3 — Perfilamiento estructural del dataset

Se revisan los tipos de datos, valores nulos, consumo de memoria y las estadísticas descriptivas de las variables numéricas.


```python
# Tipos de datos, nulos y uso de memoria
df_clientes.info()
```

```python
# Estadística descriptiva de variables numéricas
df_clientes.describe()
```

## Paso 4 — Computación vectorial con NumPy

La regla de negocio indica calcular un **Gasto Neto de Operación** aplicando una tasa del **19%** sobre el monto bruto y sumando un costo fijo logístico de **$2.5** por registro.

> **Supuesto documentado:** la guía no especifica por nombre qué columna del CSV corresponde a “montos brutos”. En el dataset entregado se utilizará `Balance` como monto monetario base. Si el docente indica otra columna, basta con reemplazar `Balance` en esta celda.


```python
# Extraer los montos brutos como arreglo NumPy
montos_brutos = df_clientes['Balance'].to_numpy()

# Parámetros de negocio
tasa_impositiva = 0.19
costo_fijo_logistico = 2.5

# Cálculo vectorizado para todos los registros
gasto_neto_operacion = (
    montos_brutos * (1 + tasa_impositiva)
    + costo_fijo_logistico
)

# Incorporar el resultado al DataFrame
df_clientes['GastoNetoOperacion'] = gasto_neto_operacion

# Verificación de los primeros resultados
df_clientes[
    ['CustomerId', 'Balance', 'GastoNetoOperacion']
].head()
```

### ¿Por qué usamos NumPy aquí?

La operación se realiza de forma **vectorizada**, es decir, sobre todos los elementos del arreglo de una sola vez, evitando recorrer manualmente cada fila con un `for`. Esto permite un procesamiento más eficiente.


## Paso 5 — Segmentación y aislamiento avanzado

La guía solicita aislar las posiciones **50 a 100 inclusive** y mantener únicamente la identificación del cliente y el nuevo gasto neto calculado.

Como el límite superior de `iloc` no se incluye, usamos `50:101`.


```python
segmento_clientes = df_clientes.iloc[
    50:101
][
    ['CustomerId', 'GastoNetoOperacion']
]

print(f"Dimensiones del segmento: {segmento_clientes.shape}")
segmento_clientes
```

## Paso 6 — Análisis de eficiencia física y exportación

Se exporta el DataFrame procesado completo a **CSV** y **Excel**, y luego se compara cuánto ocupa cada archivo en disco.


```python
# Nombres de archivos de salida
archivo_csv = 'ecommerce_customer_data_procesado.csv'
archivo_excel = 'ecommerce_customer_data_procesado.xlsx'

# Exportaciones
df_clientes.to_csv(
    archivo_csv,
    index=False,
    encoding='utf-8'
)

df_clientes.to_excel(
    archivo_excel,
    index=False
)

# Comparación de almacenamiento físico
tam_csv_bytes = os.path.getsize(archivo_csv)
tam_excel_bytes = os.path.getsize(archivo_excel)

tam_csv_mb = tam_csv_bytes / (1024 ** 2)
tam_excel_mb = tam_excel_bytes / (1024 ** 2)

print(f"Tamaño CSV:   {tam_csv_mb:.4f} MB")
print(f"Tamaño Excel: {tam_excel_mb:.4f} MB")

if tam_csv_bytes < tam_excel_bytes:
    ahorro = (1 - tam_csv_bytes / tam_excel_bytes) * 100
    print(f"CSV ocupa menos espacio que Excel ({ahorro:.2f}% menos).")
elif tam_excel_bytes < tam_csv_bytes:
    ahorro = (1 - tam_excel_bytes / tam_csv_bytes) * 100
    print(f"Excel ocupa menos espacio que CSV ({ahorro:.2f}% menos).")
else:
    print("Ambos archivos ocupan el mismo espacio.")
```
